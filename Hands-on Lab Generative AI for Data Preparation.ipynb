### Downloading the dataset

Execute the following code to download the dataset in to the interface.

> Please note that this step is essential in JupyterLite. If you are using a downloaded version of this notebook and running it on JupyterLabs, then you can skip this step and directly use the URL in pandas.read_csv() function to read the dataset as a dataframe


```python
from pyodide.http import pyfetch

async def download(url, filename):
    response = await pyfetch(url)
    if response.status == 200:
        with open(filename, "wb") as f:
            f.write(await response.bytes())

path = URL

await download(path, "dataset.csv")
file_name  = "dataset.csv"
```

```python
import pandas as pd
from typing import Union, List, Optional

class DataReadError(Exception):
    pass

def read_csv_to_dataframe(
    file_path: str,
    delimiter: str = ',',
    encoding: Optional[str] = 'utf-8',
    header: Union[int, List[int]] = 0
) -> pd.DataFrame:
    # Read a CSV file into a pandas DataFrame.
    # - header: int or list[int], e.g., 0 for first row as header, or [0, 1] for multi-row header.
    # - delimiter: field separator
    # - encoding: file encoding
    #
    # Returns the DataFrame. Raises DataReadError on failure.

    try:
        df = pd.read_csv(file_path, delimiter=delimiter, encoding=encoding, header=header)
        if df is None or df.empty:
            raise DataReadError('CSV file ' + file_path + ' is empty or contains no data rows.')
        return df
    except FileNotFoundError as e:
        raise DataReadError('CSV file not found: ' + file_path) from e
    except pd.errors.EmptyDataError as e:
        raise DataReadError('CSV file is empty: ' + file_path) from e
    except pd.errors.ParserError as e:
        raise DataReadError('Error parsing CSV file: ' + file_path) from e
    except UnicodeDecodeError as e:
        raise DataReadError('Encoding error while reading file: ' + file_path) from e
    except Exception as e:
        raise DataReadError('An error occurred while reading the CSV file: ' + file_path + '. ' + str(e)) from e

# Example usage:
df = read_csv_to_dataframe("dataset.csv")
#print(df.head())

missing_values = df.isnull().sum()
missing_columns = missing_values[missing_values > 0]
#print("Columns with missing values:")
#print(missing_columns)

most_frequent_value = df["Screen_Size_cm"].mode()[0]
df["Screen_Size_cm"] = df["Screen_Size_cm"].fillna(most_frequent_value)
mean_value = df["Weight_kg"].mean()
df["Weight_kg"] = df["Weight_kg"].fillna(mean_value)
#print("\nColumns with missing values after handling:")
#print(df.isnull().sum())


df["Screen_Size_cm"] = df["Screen_Size_cm"].astype(float)
df["Weight_kg"] = df["Weight_kg"].astype(float)
#print(df.dtypes[["Screen_Size_cm", "Weight_kg"]])


df["Screen_Size_inch"] = df["Screen_Size_cm"] / 2.54
df.drop("Screen_Size_cm", axis=1, inplace=True)
df["Weight_pounds"] = df["Weight_kg"] * 2.20462
df.drop("Weight_kg", axis=1, inplace=True)
#print(df.head())


df["CPU_frequency"] = df["CPU_frequency"] / df["CPU_frequency"].max()
#print(df["CPU_frequency"].head())


df1 = pd.get_dummies(df["Screen"], prefix="Screen")

# 2. Append df1 into the original data frame
df = pd.concat([df, df1], axis=1)

# 3. Drop the original "Screen" attribute
df = df.drop("Screen", axis=1,)
#print(df.head())


df["Price"] = df["Price"] * 0.92
df["CPU_frequency"] = (df["CPU_frequency"] - df["CPU_frequency"].min()) / \
                      (df["CPU_frequency"].max() - df["CPU_frequency"].min())
print(df.head())
```
